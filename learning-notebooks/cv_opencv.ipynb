```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat_image.webp")
cv2_imshow(image)
```

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray,(300,300))
cv2_imshow(resized)
```

```python
cropped = gray[200:600, 200:600]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
rotated = cv2.rotate(gray, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated)
rotated = cv2.rotate(gray, cv2.ROTATE_180)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 0)
cv2_imshow(flipped)
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

# Opencv video streamin codes

```python
print("Hello, World!")

import cv2

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

while True:
    success, frame = cap.read()

    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.5, minNeighbors=5)

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (255, 0, 0), 2)

    cv2.imshow('Webcam', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break


# =======================================================================

# import cv2

# cap = cv2.VideoCapture(0)

# while True:
#     success, frame = cap.read()

#     if not success:
#         break

#     gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     blurred = cv2.GaussianBlur(gray, (15, 15), 0)
#     edges1 = cv2.Canny(frame, 100, 200)
#     edges2 = cv2.Canny(frame, 150, 200)
#     flipped = cv2.flip(frame, 1)

#     # cv2.imshow('Webcam', frame)
#     # cv2.imshow('Grayscale', gray)
#     # cv2.imshow('Blurred', blurred)
#     # cv2.imshow('Edges', edges1)
#     # cv2.imshow('Edges2', edges2)
#     cv2.imshow('Flipped', flipped)


#     if cv2.waitKey(1) & 0xFF == ord('q'):
#         break

# cap.release()
# cv2.destroyAllWindows()

#=======================================================================


```
